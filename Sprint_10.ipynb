```python
# %pip install requests
```

```python
import requests
```

#### Extracción de lista de publicaciones

```python
url_posts = "https://jsonplaceholder.typicode.com/posts"

respuesta = requests.get(url_posts, timeout=30)
```

Al escribir timeout=30, es como decirle al programa: "Si pasan 30 segundos sin recibir datos del servidor, deja de esperar y muestra un error"

```python
publicaciones = respuesta.json()

display(publicaciones[:2])
```

#### Extracción de listas de usuarios

```python
url_usuarios = "https://jsonplaceholder.typicode.com/users"

respuesta_usuarios = requests.get(url_usuarios, timeout=30)

usuarios = respuesta_usuarios.json()

display(usuarios[:2])
```

#### Extracción de listas de tareas

```python
url_tareas = "https://jsonplaceholder.typicode.com/todos"

respuesta_tareas = requests.get(url_tareas, timeout=30)

tareas = respuesta_tareas.json()

display(tareas[:2])
```

### Pregunta 5. Realiza una petición PATCH para modificar parcialmente una publicación existente.

* Muestra la respuesta JSON.
* Muestra el código de estado.

```python
cambios = {
    "title": "Título modificado por Lari"
}
```

```python
respuesta_patch = requests.patch(
    f"{url_posts}/1",
    json=cambios,
    timeout=30
)
```

```python
display(respuesta_patch.json())

print("Código de estado:", respuesta_patch.status_code)
```

Esto es una modificación parcial: se ha enviado un cambio al título, pero los campos "body" y "userId" conservan sus valores en la respuesta. 

Se ha usado la publicación 1, que ya existe, porque la 101 que se creó anteriormente, fue una creación simulada y no quedó guardada. Aquí la modificación se simula.

**API SELECCIONADA**

He elegido Open-Meteo para consultar la previsión de lluvia de Barcelona durante los próximos 7 días mediante peticiones GET.

**Endpoints**

1. https://geocoding-api.open-meteo.com/v1/search
   Permite buscar localidades por su nombre y obtener información
   como sus coordenadas y país.

2. https://api.open-meteo.com/v1/forecast

**Parámetros**

Para buscar localidades:

- name: nombre de la localidad. Es obligatorio.
- count: cantidad máxima de resultados. Es opcional.
- language: idioma de los resultados. Es opcional.
- format: formato de respuesta; utilizaremos JSON.

Para consultar la previsión:

- latitude y longitude: coordenadas de la ubicación. Son obligatorias.
- daily: selecciona las variables diarias.
  Utilizaremos rain_sum para obtener la lluvia prevista acumulada por día.
- forecast_days: número de días de previsión. Es opcional;
  utilizaremos 7.
- timezone: zona horaria. Es necesaria al solicitar datos diarios;
  utilizaremos Europe/Madrid.

**Formato de respuesta**

La API devuelve los datos en formato JSON.

### 3. Haz una petición GET sencilla:

Muestra el código de estado de la respuesta.
Imprime de forma clara algunos de los campos de la respuesta JSON.

```python
url_localidades = "https://geocoding-api.open-meteo.com/v1/search"

parametros = {
    "name": "Barcelona",
    "count": 1,
    "language": "es",
    "format": "json"
}

respuesta_localidades = requests.get(
    url_localidades,
    params=parametros,
    timeout=30
)

print("Código de estado:", respuesta_localidades.status_code)
```

```python
datos_localidades = respuesta_localidades.json()

barcelona = datos_localidades["results"][0]

print("Localidad:", barcelona["name"])
print("País:", barcelona["country"])
print("Latitud:", barcelona["latitude"])
print("Longitud:", barcelona["longitude"])
```

Una vez obtenidas estas coordenadas, de Barcelona (España), las utilizaremos para pedir la previsión de lluvia de 7 días.

```python
url_prevision = "https://api.open-meteo.com/v1/forecast"

parametros_prevision = {
    "latitude": barcelona["latitude"],
    "longitude": barcelona["longitude"],
    "daily": "rain_sum",
    "forecast_days": 7,
    "timezone": "Europe/Madrid"
}

respuesta_prevision = requests.get(
    url_prevision,
    params=parametros_prevision,
    timeout=30
)

print("Código de estado:", respuesta_prevision.status_code)
```

```python
datos_prevision = respuesta_prevision.json()

display(datos_prevision["daily"])
```

### 4. Convierte los datos a un DataFrame de pandas:

Muestra las primeras filas de DataFrame.

```python
import pandas as pd

df_prevision = pd.DataFrame(datos_prevision["daily"])

display(df_prevision)
```

## NIVEL 3: API de Open Data Barcelona

### 1. Utiliza la API de Open Data BCN para buscar un conjunto de datos de tu interés mediante package_search y package_show.

| Acción | Para qué sirve |
|---|---|
| `package_search` | Buscar conjuntos de datos dentro del catálogo. |
| `package_show` | Consultar la ficha de un conjunto concreto, incluyendo sus recursos. |

```python
url_busqueda = (
    "https://opendata-ajuntament.barcelona.cat"
    "/data/api/3/action/package_search"
)

parametros_busqueda = {
    "q": "meteo",
    "rows": 5
}

respuesta_busqueda = requests.get(
    url_busqueda,
    params=parametros_busqueda,
    timeout=30
)

print("Código de estado:", respuesta_busqueda.status_code)
```

```python
datos_busqueda = respuesta_busqueda.json()

print("Búsqueda realizada:", datos_busqueda["success"])
```

```python
datasets = datos_busqueda["result"]["results"]

print("Total de coincidencias:", datos_busqueda["result"]["count"])

for dataset in datasets:
    print("Título:", dataset["title"])
    print("Identificador:", dataset["name"])
    print()
```

Usaré el segundo conjunto de datos, el de mesures-estacions-meteorologiques, ya que por su título, contiene recursos estadísticos de las estaciones meteorológicas.

```python
url_ficha = (
    "https://opendata-ajuntament.barcelona.cat"
    "/data/api/3/action/package_show"
)

parametros_ficha = {
    "id": "mesures-estacions-meteorologiques"
}

respuesta_ficha = requests.get(
    url_ficha,
    params=parametros_ficha,
    timeout=30
)

print("Código de estado:", respuesta_ficha.status_code)
```

```python
datos_ficha = respuesta_ficha.json()

print("Consulta realizada:", datos_ficha["success"])
```

```python
ficha = datos_ficha["result"]

print("Título:", ficha["title"])
print("Número de recursos:", len(ficha["resources"]))
```

### 2. De los resultados obtenidos, selecciona uno que tenga recursos disponibles en CSV o JSON.

Anotación a tener en cuenta:

El primer recurso, 2026_MeteoCat_Detall_Estacions.csv, tiene formato CSV y DataStore activo: True. Así que lo guardamos en una variable.

```python
recurso_elegido = ficha["resources"][0]

resource_id = recurso_elegido["id"]

print("Recurso elegido:", recurso_elegido["name"])
print("ID:", resource_id)
```

### 3. Con el resource_id del recurso seleccionado, realiza una consulta mediante datastore_search para recuperar al menos 100 registros.

```python
url_registros = (
    "https://opendata-ajuntament.barcelona.cat"
    "/data/api/3/action/datastore_search"
)

parametros_registros = {
    "resource_id": resource_id,
    "limit": 100
}

respuesta_registros = requests.get(
    url_registros,
    params=parametros_registros,
    timeout=30
)

print("Código de estado:", respuesta_registros.status_code)
```

```python
datos_registros = respuesta_registros.json()

print("Consulta realizada:", datos_registros["success"])
```

```python
registros = datos_registros["result"]["records"]

print("Registros recibidos:", len(registros))

display(registros[:2])
```

### 4. Convierte los resultados en un DataFrame.

```python
df_meteo = pd.DataFrame(registros)

display(df_meteo)
```

Se han recuperado 100 registros del recurso meteorológico de 2026 y se han convertido en un DataFrame. Cada fila incluye el código de la estación, la fecha de lectura, la abreviatura de una variable y su valor.

Se observa un problema de codificación en el nombre de la columna de abreviaturas y la presencia del texto «NA» en algunas celdas de DATA_EXTREM, cuyo significado debe comprobarse en la documentación.

Estos 100 registros constituyen una extracción limitada del recurso. Antes de realizar un análisis meteorológico, sería necesario consultar la descripción y las unidades de cada variable, revisar los tipos de datos y determinar el alcance temporal y territorial de la extracción.

### 5. Guarda el DataFrame en un archivo .csv.

```python
df_meteo.to_csv(
    "datos_meteorologicos_2026.csv",
    index=False,
    encoding="utf-8-sig"
)
```

`index=False`: evita guardar el índice de Pandas —los números 0, 1, 2…— como una columna adicional. La columna _id sí se conserva.

`encoding="utf-8-sig`": utiliza una codificación que facilita que Excel reconozca los caracteres. No repara los textos que ya llegaron mal codificados.
